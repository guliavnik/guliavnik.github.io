Для пропусков в столбцах `critic_score`, `user_score` и `eu_sales` мы найдем средние значения, с учетом платформы игры и и года ее релиза. Пропуски в столбцах `name` и `genre` мы заменим пустой строкой. 

Пропуски для столбоца  `jp_sales` будут удалены, потому что их процент крайне незначителен.
Проспуски в `year_of_release` будут удалены, потому что они помешают фильтрации по годам, а их процент небольшой, заполнить недостающие значения используя имеющиеся данные мы не можем. Пропуски в столбце `age_rating` мы оставим без вниамния, так как эти данные не нужны нам для анализа.

- Обработайте пропущенные значения. Для каждого случая вы можете выбрать оптимальный, на ваш взгляд, вариант: заменить на определённое значение, оставить как есть или удалить.
- Если вы решите заменить пропуски на значение-индикатор, то убедитесь, что предложенное значение не может быть использовано в данных.
- Если вы нашли пропуски в данных с количеством проданных копий игры в том или ином регионе, их можно заменить на среднее значение в зависимости от названия платформы и года выхода игры.

```python
df = df.dropna(subset=['year_of_release'])
df = df.dropna(subset=['jp_sales']) # удаляем строки с просками в годе релиза и продажах в Японии
```

```python
df[['name','genre']] = df[['name','genre']].fillna(' ') # меняем пропуски в названии игры и названии жанра на пробел
```

```python
def mean_critic_score(r):
    if pd.isna(r['critic_score']):
        filtred = df[(df['year_of_release'] == r['year_of_release']) & 
                     (df['platform'] == r['platform'])]
        return filtred['critic_score'].mean()
    else: 
        return r['critic_score'] # создаем функцию для подсчета средних значений с учетом года релиза и 
                                 # платформы для оценок критиков    
    
def mean_user_score(r):
    if pd.isna(r['user_score']):
        filtred = df[(df['year_of_release'] == r['year_of_release']) & 
                     (df['platform'] == r['platform'])]
        return filtred['user_score'].mean() 
    else: 
        return r['user_score'] # создаем функцию для подсчета средних значений с учетом года релиза и 
                               # платформы для оценок пользователей
def mean_eu_sales(r):
    if pd.isna(r['eu_sales']):
        filtred = df[(df['year_of_release'] == r['year_of_release']) & 
                     (df['platform'] == r['platform'])]
        return filtred['eu_sales'].mean()
    else: 
        return r['eu_sales'] # создаем функцию для подсчета средних значений с учетом года релиза и 
                             # платформы для продаж в европе
    
df['critic_score'] = df.apply(mean_critic_score, axis=1) 
df['user_score'] = df.apply(mean_user_score, axis=1)
df['eu_sales'] = df.apply(mean_user_score, axis=1) # последовательно применяем функции
```

```python
round(df.isna().sum() / len(df) * 100, 2) # проверяем результат
```

```python
df = df.dropna(subset=['critic_score'])
df = df.dropna(subset=['user_score'])
df = df.dropna(subset=['eu_sales'])
round(df.isna().sum() / len(df) * 100, 2) # удаляем пропуски, которые не удалось заполнить
```

Мы заменили на средние значения пропуски в столбцах `critic_score`, `user_score` и `eu_sales` и удалили значения, которые не удалось заполнить.

Пропуски в столбцах `name` и `genre` мы замениили пустой строкой, а пропуски в `jp_sales` удалили.
